```python
import pandas as pd
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Flatten, Input, Dropout
from sklearn.model_selection import train_test_split
from gensim.parsing.preprocessing import preprocess_string
from nltk import sent_tokenize
```

```python
df = pd.read_csv('data_mr.csv').iloc[5000:20000]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['text'] = df['text'].apply(preprocess_string)
```

```python
df
```

```python
y=df['label']
```

```python
X=df['text']
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.2,random_state=7)
X_train = [" ".join(doc) for doc in X_train]
X_test = [" ".join(doc) for doc in X_test]
```

```python
X_train[0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv_obj=CountVectorizer()
```

```python
X_train_bow = cv_obj.fit_transform(X_train).toarray()
X_test_bow = cv_obj.transform(X_test).toarray()
```

```python
nn = Sequential([
    Dense(128, activation='relu', input_shape=(36318,)),
    Dropout(0.3),

    Dense(64, activation='relu'),
    Dropout(0.3),

    Dense(1, activation='sigmoid')
])
```

```python
from tensorflow.keras.optimizers import Adam
nn.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
nn.fit(X_train_bow,y_train,epochs=15)
```

```python
nn.evaluate(X_test_bow,y_test)
```

```python
from sklearn.metrics import f1_score

y_pred = (nn.predict(X_test_bow) > 0.5).astype(int)
print(f1_score(y_test, y_pred))
```
